### Shannon Entropy equation:
$$ H = -\sum_{i \in \{A,C,G,U\}} p(i) \cdot \log_2 p(i) $$
where $p(i)$ = frequency of nucleotide $i$ in the sequence. $H = 0$ for a pure homopolymer (e.g. poly-U), $H = 2$ for equal mix of all 4 nucleotides.

```python
#feature extraction, followed their TerFac-offline git repository
from collections import Counter
import math

def shannon_entropy(seq):
    if not seq:
        raise ValueError("seq is empty or None")
    length = len(seq)
    counts = Counter(seq)
    entropy = 0.0
    for nt, count in counts.items():
        p = count/length
        if p > 0:
            entropy -= p*math.log2(p)
    return round(entropy / 2, 4) 
    # here did the same as their TerFac-offline code, divide by 2 to normalize the entropy to [0,1] range, since max entropy for 4 symbols is log2(4)=2.
    # and keep 4 dp
    
```

```python
def state_changes(seq):
    if not seq:
        raise ValueError("seq is empty or None")

    return sum(1 for i in range(len(seq)-1) if seq[i] != seq[i+1]) #do not normalized for now
    # return round(changes / (len(seq) - 1), 4) #here do the same as their TerFac-offline code
```

```python
def parse_hairpin(hairpin_seq, structure):
    """Split hairpin into stem (paired only) and apical loop.

    Bulges and internal loops are discarded. Only the innermost
    unpaired region (between last '(' and first ')' after it) is
    returned as the apical loop. Assume there is only 1 apical loop exist.

    TODO: For now, we assume the sequence and structure can match
        with each other. If needed, will change later.

    Args:
        hairpin_seq: RNA sequence string (e.g. "GACGCUGAAAAGCGUC").
        structure: Dot-bracket notation string, same length as
            hairpin_seq. '(' and ')' denote paired positions,
            '.' denotes unpaired.

    Returns:
        A tuple of (stem, apical_loop) where stem is the concatenation
        of all paired nucleotides and apical_loop is the sequence
        between the innermost base pair. Bulge and internal loops are removed.

    Raises:
        ValueError: If hairpin_seq and structure have different lengths.
    """
    if len(hairpin_seq) != len(structure):
        raise ValueError(f"length mismatch: seq={len(hairpin_seq)}, struct={len(structure)}")
    if '(' not in structure:
        raise ValueError(f"No hairpin found in structure: {structure}")
    # Find apical loop: dots between last '(' and first ')' after it
    last_open = structure.rindex('(') #find the last occurance of a specific value
    first_close = structure.index(')', last_open) #start search from last open, return is with respect to the full length of the string

    # print(last_open)
    # print(first_close)
    
    # Apical loop = everything between last '(' and first ')'
    apical_loop = hairpin_seq[last_open + 1 : first_close]
    
    # Stem = only paired nucleotides (brackets), no dots
    stem = ""
    for nt, bracket in zip(hairpin_seq, structure):
        if bracket in ('(', ')'):
            stem += nt
    
    return stem, apical_loop

# test
hairpin = "CCGCCGGUCUUGUCCACUACCUUGCAGUAAUGCGGUGGACAGGAUCGGCGG"
struct  = "((((((((((((((((((.....(((....)))))))))))))))))))))"
stem, loop = parse_hairpin(hairpin, struct)
print("Stem:", stem)
print("Apical loop:", loop)
print("Stem length:", len(stem))
```

```python
# test cases for parse_hairpin

# 1. Simple hairpin, no bulges (pheA-1 terminator)
stem, loop = parse_hairpin("GACGCUGAAAAGCGUC", "((((((....))))))") 
assert stem == "GACGCUAGCGUC", f"stem={stem}"
assert loop == "GAAA", f"loop={loop}"

# 2. Complex hairpin with internal unpaired regions (bulges)
stem, loop = parse_hairpin(
    "CCGCCGGUCUUGUCCACUACCUUGCAGUAAUGCGGUGGACAGGAUCGGCGG",
    "((((((((((((((((((.....(((....)))))))))))))))))))))"
)
assert loop == "GUAA", f"loop={loop}"
assert len(stem) == 42

# 3. Long loop (6-nt, thrLABC terminator)
stem, loop = parse_hairpin("GCCCGCACCUGACAGUGCGGGC", 
                           "((((((((......))))))))")
assert stem == "GCCCGCACGUGCGGGC", f"stem={stem}"
assert loop == "CUGACA", f"loop={loop}"

# 4. 5-nt loop (secG-leuU terminator)
stem, loop = parse_hairpin("GGCCUCCCAAAUCGGGGGGCC", 
                           "((((((((.....))))))))")
assert loop == "AAAUC", f"loop={loop}"
assert stem == "GGCCUCCCGGGGGGCC", f"stem={stem}"

# --- corner cases ---

# 5. Minimal 3-nt loop
#    seq:    G C G A U A C G C
#    struct: ( ( ( . . . ) ) )
#    loop = seq[3:6] = "AUA", stem = positions 0,1,2,6,7,8 = "GCGCGC"
stem, loop = parse_hairpin("GCGAUACGC", "(((...)))")
assert loop == "AUA", f"loop={loop}"
assert stem == "GCGCGC", f"stem={stem}"

# 6. Multiple bulges — stem should exclude all unpaired nucleotides
#    seq:    G  C  A  U  U  G  C  A  A  A  G  C  A  A  U  G  C
#    struct: (  (  .  .  (  (  (  .  .  .  )  )  )  .  .  )  )
#    loop = seq[7:10] = "AAA", stem = indices 0,1,4,5,6,10,11,12,15,16 = "GCUGCGCAGC"
stem, loop = parse_hairpin("GCAUUGCAAAGCAAUGC", "((..(((...)))..))") 
assert loop == "AAA", f"loop={loop}"
assert stem == "GCUGCGCAGC", f"stem={stem}"

# 7. Length mismatch should raise ValueError
try:
    parse_hairpin("GACGCU", "(((...))")
    assert False, "should have raised ValueError"
except ValueError:
    pass

print("All tests passed!")
```

```python
def initial_gc_count(sequence):
    """
    Count how many consecutive nucleotides at the start of the sequence are either G or C.
    """
    count = 0
    for nt in sequence:
        if nt in ['C', 'G']:
            count += 1
        else:
            break
    return count
```

```python
"""
Test cases for feature extraction (v2 - raw state_changes).
Paste this into a new Jupyter cell and run it.
All assertions should pass silently — if something is wrong, you'll see an AssertionError.
"""


# === Make sure your helper functions are already defined above ===

def test_shannon_entropy():
    """Test Shannon entropy edge cases and known values"""
    assert shannon_entropy("UUUUUUUUUUUU") == 0.0, "All-U should be 0"
    assert shannon_entropy("AAAAAAAA") == 0.0, "All-A should be 0"
    assert shannon_entropy("ACGU") == 1.0, "Equal ACGU should be 1.0"
    assert shannon_entropy("ACGUACGU") == 1.0, "Equal 2xACGU should be 1.0"
    assert shannon_entropy("AUAU") == 0.5, "Equal AU should be 0.5"
    print("✓ shannon_entropy passed all tests")

def test_state_changes():
    """Test raw state change count (NOT normalized)"""
    assert state_changes("UUUUUU") == 0, "All same = 0"
    assert state_changes("AAAAAAAA") == 0, "All A = 0"
    assert state_changes("AUAUAU") == 5, "Alternating 6-mer = 5"
    assert state_changes("UUUUUUCUUUUC") == 3, "UUUUUUCUUUUC = 3 changes"
    assert state_changes("AGAAAAGA") == 4, "AGAAAAGA = 4 changes"
    print("✓ state_changes passed all tests")

def test_initial_gc_count():
    """Test GC initial count"""
    assert initial_gc_count("GACGCU") == 1, "G then A → 1"
    assert initial_gc_count("GCCGGG") == 6, "All GC → 6"
    assert initial_gc_count("AGCGC") == 0, "Starts with A → 0"
    assert initial_gc_count("CCGGCA") == 5, "CCGGC then A → 5"
    assert initial_gc_count("GGCCUC") == 4, "GGCC then U → 4"
    print("✓ initial_gc_count passed all tests")

def test_parse_hairpin():
    """Test hairpin parsing"""
    # Simple case
    stem, loop = parse_hairpin("GACGCUGAAAAGCGUC", "((((((....))))))")
    assert stem == "GACGCUAGCGUC", f"Simple stem wrong: {stem}"
    assert loop == "GAAA", f"Simple loop wrong: {loop}"
    
    # Bulge case
    stem2, loop2 = parse_hairpin("AAGCGCCAUUCAGCGCCU", 
                                 "(.((((......)))).)") 
    assert stem2 == "AGCGCGCGCU", f"Bulge stem wrong: {stem2}"
    assert loop2 == "CAUUCA", f"Bulge loop wrong: {loop2}" #this one claude wrote it wrong and I fixed it
    
    # Internal loop case
    stem3, loop3 = parse_hairpin("GGCCCCACUUUCGUGGAGCC", "(((.((((....)))).)))")
    assert len(stem3) < len("GGCCCCACUUUCGUGGAGCC"), "Should remove unpaired nts"
    
    print("✓ parse_hairpin passed all tests")

def test_feature_values_row0():
    """Verify all 16 features for Row 0 (L3S2P33)"""
    u = "UUUUUUCUUUUC"
    a = "AGAAAAGA"
    loop = "GAAA"
    hp = "GACGCUGAAAAGCGUC"
    st = "((((((....))))))"
    min_l, max_l = 3, 16
    
    # U-tract
    assert round(u[:10].count('U')/10, 4) == 0.9
    assert round(u[:6].count('U')/6, 4) == 1.0
    assert round(u[:6].count('A')/6, 4) == 0.0
    assert round(u.count('C')/12, 4) == 0.1667
    assert round(state_changes(u) / (len(u)-1), 4) == 0.2727, "U-tract SC normalized"
    assert shannon_entropy(u) == 0.325
    
    # A-tract
    assert shannon_entropy(a) == 0.4056
    assert round(state_changes(a) / (len(a)-1), 4) == 0.5714, "A-tract SC normalized"
    assert round(a[-6:].count('A')/6, 4) == 0.8333
    assert round(a[-6:].count('C')/6, 4) == 0.0
    
    # Loop
    assert round((len(loop)-min_l)/(max_l-min_l), 4) == 0.0769
    assert round((loop.count('G')+loop.count('C'))/len(loop), 4) == 0.25
    
    # Hairpin
    stem, _ = parse_hairpin(hp, st)
    first = stem[:len(stem)//2]
    assert shannon_entropy(stem) == 0.9591
    assert len(first)*2 == 12
    assert state_changes(first) * 2 + 1 == 11, "Hairpin SC raw: 5*2+1=11"
    assert initial_gc_count(first) == 1
    
    print("✓ Row 0 (L3S2P33) all 16 features verified")

def test_feature_values_row7():
    """Edge case: all-A A-tract (ECK120010855-R)"""
    a = "AAAAAAAA"
    assert shannon_entropy(a) == 0.0
    assert state_changes(a) == 0
    assert round(a[-6:].count('A')/6, 4) == 1.0
    
    # Hairpin: CCGGCGCAAUGGCCGG with structure (((((......)))))
    hp = "CCGGCGCAAUGGCCGG"
    st = "(((((......)))))"
    stem, loop = parse_hairpin(hp, st)
    first = stem[:len(stem)//2]
    assert initial_gc_count(first) == 5, "CCGGC starts with 5 G/C"
    
    print("✓ Row 7 (ECK120010855-R) edge case verified")

def test_feature_values_row8():
    """Row 8: weak terminator with high A in U-tract"""
    u = "UAAUACCGAUUA"
    assert round(u[:10].count('U')/10, 4) == 0.3
    assert round(u[:6].count('U')/6, 4) == 0.3333
    assert round(u[:6].count('A')/6, 4) == 0.5
    assert round(u.count('C')/12, 4) == 0.1667
    assert state_changes(u) == 8, "UAAUACCGAUUA raw changes = 8"
    assert round(state_changes(u) / 11, 4) == 0.7273, "normalized = 8/11"
    
    print("✓ Row 8 (ECK120030801) verified")

def test_hairpin_sc_variable_length():
    """Verify hairpin SC uses raw count, not normalized — critical for variable-length stems"""
    # Two stems with same raw changes but different lengths
    short = "GACGCU"    # len=6, raw_sc=5
    long_ = "GACGCUACGU" # len=10, raw_sc=9
    
    # Raw counts should differ
    sc_short = state_changes(short) * 2 + 1  # 5*2+1 = 11
    sc_long = state_changes(long_) * 2 + 1   # 9*2+1 = 19
    
    assert sc_short == 11
    assert sc_long == 19
    
    # If we had normalized first (the old bug), both would give different wrong values
    # This test confirms we're using raw counts
    print("✓ Hairpin SC variable-length test passed")

# === RUN ALL TESTS ===
test_shannon_entropy()
test_state_changes()
test_initial_gc_count()
test_parse_hairpin()
test_feature_values_row0()
test_feature_values_row7()
test_feature_values_row8()
test_hairpin_sc_variable_length()
print("\n🎉 ALL TESTS PASSED")
```
